# Preparation of results for F2adef -- the rest of Proof-of-Concept study

This notebook summarizes the data computed by ../scripts/F2ade.sh and ../scripts/F2f.sh.

```python
import os
import os.path as op
import pandas as pd

# Directory containing the processed F2 data
data_dir = "../data/processed/F2/"

# Get the list of files in the directory
files = [a for a in os.walk(data_dir)][0][2]

# Read each file as a DataFrame
dataframes = [
    pd.read_csv(op.join(data_dir, b), sep="\t", index_col=0)
    for b in files
]

# Concatenate all DataFrames and write the result to a TSV file
pd.concat(dataframes).to_csv(
    "../data/processed/F2bcdef.tsv",
    sep="\t"
)
```
